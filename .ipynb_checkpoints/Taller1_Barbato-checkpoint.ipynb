# Filtrado de base de datos: 
El EDA se enfocará en los datos de terremotos en Sur América, para ésto debemos filtrar la base de datos con ciertos criterios.

# Identificamos los paises que componen Sur América:

 



Filtrando:

```python
paises_margen=["Chile", "Peru", "Ecuador", "Colombia"] #Lista de paises que componen el márgen sur americano.
paises_no_margen=["Argentina", "Uruguay", "Bolivia", "Venezuela", "Guyana", "Brazil", "Paraguay",] #Se define también una lista con los otros paises SA.
paises_margen_copia=paises_margen.copy() #Creamos una copia para poder trabajarla sin alterar la lista paises_margen, la cual se utilizará nuevamente.
paises_margen_copia.extend(paises_no_margen) #Extendemos la lista paises_margen_copia con la fn extend.
paises_sudamerica=paises_margen_copia #Reasignamos el nombre de la lsita extendida.
```

## Estableciendo LCR (Longitud Cartesiana de Referencia):
* la lista paises_margen se utilizará para sacar una longitud promedio cartesiana (LCR), con el fin de obtener una referencia para las posteriores conclusiones relacionadas a la distribución geográfica de los datos.

```python
df_t=df_t.dropna(subset=["place"])#Redefinimos df_t sin los valores NaN
df_paises_m=df_t[df_t["place"].str.contains("|".join(paises_margen))]#Filtramos el DF en uno en el que solo contenga las str que contiene paises_margen.
l_pm=df_paises_m["longitude"] # DF solo con longitud.
type(l_pm)
LCR=l_pm.mean() #promedio de la longitud (cartesiana) de los paises del márgen sur americano.
LCR
```

Ya que las ciudades (con nombres iguales a de los paises del margen) se terminan y separa con "coma espacio": ", ". Se puede utilizar este patrón para filtrar de mejor manera los datos.

```python
df_paises_m=df_t[df_t["place"].str.split(", ").str[-1].isin(paises_margen)]
#Reevaluamos LCR, ahora de manera más precisa gracias al filtro anterior:
l_pm=df_paises_m["longitude"] #l_pm: longitud piases margen (dataframe)
LCR = round(l_pm.mean(),3)
LCR
```

* LCR=-73.473

La linea anaranjada representa la Longitud Cartesiana de Referencia, LCR.

## Medidas de tendencia central.

Se utilizará el DataFrame filtrado con los paises sudamericanos para calcular mediana, media y moda

```python
#Creamos df de los paises sudamericanos, dando como parámetro la lista paises_sudamerica
df_paises_s=df_t[df_t["place"].str.split(", ").str[-1].isin(paises_sudamerica)]

df_sud_num=df_paises_s.select_dtypes(include=[float,int]) #Seleccionamos solo daton numéricos.

#Media:
media_df_sud=df_sud_num.mean()
#Moda:
moda_df_sud=df_sud_num.mode()
#Mediana:
mediana_df_sud=df_sud_num.median()

print(media_df_sud)

print(moda_df_sud)

print(mediana_df_sud)

```

## Conclusiones ##

De acuerdo a los histogramas obtenidos, es posible caracterizar de forma más o menos simplificada el continente sur americano, en relación a sus particularidades geodinámicas y estructurales, derivadas e inferidas a partir de dichos resultados.

* Distribución geográfica de los terremotos:

De acuerdo a los histogramas de latitud y longitud es posible inferir que **no** en todo el continente ocurren sismos con la misma frecuencia, sino que existe un rango de longitud y latitud en la cual los datos son más abundantes. Mientras que la longitud presenta una desviación típica de 4.61 , donde los datos rondan en torno a los 70° longitud, la desviación típica de la latitud es de 13.93, con más frecuencias similares y con una moda poco desigual con respecto a los demás datos, rondando en torno a los -25° de latitud. Ésto nos describe una zona longitudinal N-S, donde los terremotos han sido mayormente datados. Es posible entonces relacionar esta "zona estadística" con la bibliografía, resultando en una gran concordancia con el límite entre las placas de Nazca y Sur America, lugar donde ocurre la subducción y la mayoría de los terremotos de las mayores magnitudes datadas.

* Relación entre magnitud y distancia con la fosa:

Anteriormente se definió LCR. Sobre esta longitud de referencia es donde se concentran la mayor cantidad de terremotos y de las mayores magnitudes. Alejándonos perpendicularmente de ésta línea, es posible observar (en el gráfico de multivariable Longitud vs Magnitud) una relación inversa, sustentada también por la co-varianza obtenida (-0.1075) la cual es menos a cero. Si bien la relación inversa no es muy intensa (posiblemente debido a las distancias involucradas en el estudio), es posible inferir que a medida que nos alejamos de la zona de subducción, la abundancia y magnitud de los terremotos decrece. Se sugiere que a medida nos adentramos en el continente, nos alejamos de la ventana sísmica, por lo que sus consecuencias, principalmente en forma de ondas sísmicas, no son tan importantes.

* Datos atípicos en diferentes paises:

De acuerdo al gráfico de boxplot:
1. Existen paises donde los datos son muy excasos, inculuso cero: Uruguay, Brazil, Paraguay, Guyana. 
2. Otros Paises, como Chile, Ecuador y Perú muestran una gran densidad de datos fuera de los cuartiles: Esto posiblemente responde a que son paises que se encuentran cerca del márgen sudamericano, en donde los terremotos son más fuertes y escapan de las tendencias centrales de magnitud. Casos como el De Valdivia 1960 y Maule 2010 son registros de datos atípicos ocurridos en Chile. Se concluye que en estos paises es donde existen la mayor cantidad y los terremotos más fuertes del continente, ya que poseen una gran cantidad de registros atípicos, donde contrarestan con la media de las magnitudes que es 4.93.

```python

import matplotlib.pyplot as plt 
```

```python
filtro1=(df_t['mag']>=5.5) & (df_t['depth'] <150)
terremotos_b100=df_t[filtro1].dropna()
terremotos_b100
terremotos_b100['fecha'] = pd.to_datetime(terremotos_b100['time']).dt.strftime('%Y-%m')
terremotos_b100['año'] = pd.to_datetime(terremotos_b100['time']).dt.year
terremotos_b100['mes'] = pd.to_datetime(terremotos_b100['time']).dt.month

filtro2=(terremotos_b100['año'] >=2022) 
terremotos_b100k_m2000t=terremotos_b100[filtro2]

terremotos_b100k_m2000t['fecha'] = pd.to_datetime(terremotos_b100k_m2000t['fecha'])

terremotos_b100k_m2000t['mes_acumulado'] = (terremotos_b100k_m2000t['fecha'].dt.year - 2022) * 12 + terremotos_b100k_m2000t['fecha'].dt.month
            
 
plt.figure(figsize=(14, 6))

plt.hist(terremotos_b100k_m2000t['mes_acumulado'],bins=10,edgecolor='black')
plt.xticks(ticks=range(4,15), labels=[str(i+1) for i in range(4,15)])



plt.title('Sismos >5.5[mag] & <150km[depth]')
plt.text(10,54,'*Desde 2022*',fontsize='x-large')
plt.text(3.3,-7.5,'2022-04',rotation=45,fontsize='medium')
plt.text(4.3,-7.5,'2022-05',rotation=45,fontsize='medium')
plt.text(5.3,-7.5,'2022-06',rotation=45,fontsize='medium')
plt.text(6.3,-7.5,'2022-07',rotation=45,fontsize='medium')
plt.text(7.3,-7.5,'2022-08',rotation=45,fontsize='medium')
plt.text(8.3,-7.5,'2022-09',rotation=45,fontsize='medium')
plt.text(9.3,-7.5,'2022-10',rotation=45,fontsize='medium')
plt.text(10.3,-7.5,'2022-11',rotation=45,fontsize='medium')
plt.text(11.3,-7.5,'2022-12',rotation=45,fontsize='medium')
plt.text(12.3,-7.5,'2023-01',rotation=45,fontsize='medium')
plt.text(13.3,-7.5,'2023-02',rotation=45,fontsize='medium')
plt.xlabel('tiempo [meses]',labelpad=40)
plt.ylabel('Frecuencia absoluta')
plt.xlim(3,15)
plt.grid(True)


plt.show
```

Creamos el gráfico de función de probabilidad para los sismos que cumplen nuestra condicón, transformamos los datos de tiempo para agruparlos en meses desde el 4/2022 (mes 4) al 2/2023 (mes 14).

```python
plt.hist(terremotos_b100k_m2000t['mes_acumulado'],density=True)
```

```python
import scipy.stats
from scipy.stats import randint, logser, poisson, hypergeom, geom, nbinom, zipf, binom

plt.figure(figsize=(14, 8))

plt.hist(terremotos_b100k_m2000t['mes_acumulado'],density=True, alpha=0.5)
plt.xticks([])

plt.title('Sismos >5.5[mag] & <150km[depth]')
plt.xlabel('tiempo [meses]',labelpad=8)
plt.ylabel('Probabilidad')

#1. distribución binomial

x = np.arange(4, 15)
dist_binom=binom.pmf(x,14,0.6)
plt.plot(x,dist_binom,label='Distribución binomial')

#2. distribución uniforme
dist_unif = randint.pmf(x, 5, 14)
plt.plot(x,dist_unif, label='Distribución uniforme')

#3. distribución de logarítmica
dist_bern = logser.pmf(x, 0.8)
plt.plot(x,dist_bern,label='Distribución logarítmica')

#4. distribución de poisson
dist_poiss= poisson.pmf(x, 7)

plt.plot(x,dist_poiss,label='Distribución Poisson')

#5. distribución hipergeométrica
dist_hyper = hypergeom.pmf(x, 300, 13, 80)
plt.plot(x,dist_hyper, label='Distribución hipergeométrica')

#6. Distribución geométrica
dist_geom = geom.pmf(x, 0.03)
plt.plot(x,dist_geom, label='Distribución geométrica')

#7. Distribución binomial negativa
dist_nbinom = nbinom.pmf(x, 14, 0.52)
plt.plot(x,dist_nbinom, label='Distribución binomial negativa')

#8. Distribución de zipf

dist_zipf = zipf.pmf(x,3)
plt.plot(x,dist_zipf, label='Distribución zipf')
plt.text(3.3,-0.03,'2022-04',rotation=45,fontsize='medium')
plt.text(4.3,-0.03,'2022-05',rotation=45,fontsize='medium')
plt.text(5.3,-0.03,'2022-06',rotation=45,fontsize='medium')
plt.text(6.3,-0.03,'2022-07',rotation=45,fontsize='medium')
plt.text(7.3,-0.03,'2022-08',rotation=45,fontsize='medium')
plt.text(8.3,-0.03,'2022-09',rotation=45,fontsize='medium')
plt.text(9.3,-0.03,'2022-10',rotation=45,fontsize='medium')
plt.text(10.3,-0.03,'2022-11',rotation=45,fontsize='medium')
plt.text(11.3,-0.03,'2022-12',rotation=45,fontsize='medium')
plt.text(12.3,-0.03,'2023-01',rotation=45,fontsize='medium')
plt.text(13.3,-0.03,'2023-02',rotation=45,fontsize='medium')

plt.legend()

```

```python
media1 = terremotos_b100k_m2000t['mes_acumulado'].mean()
varianza1 = terremotos_b100k_m2000t['mes_acumulado'].var()
print(f'media: {media1:.2f}, varianza: {varianza1:.2f}')
```

Prueba de hipótesis

```python
# Calcular valores esperados usando distribución randint
a, b = 5, 14  # Parámetros para la distribución randint
valores_esperados_u = [randint.pmf(x, a, b) for x in terremotos_b100k_m2000t['mes_acumulado']]

# Convertir a array para el cálculo del estadístico chi-cuadrado
observados = np.array(terremotos_b100k_m2000t['mes_acumulado'])
esperados = np.array(valores_esperados_u)

# Calcular el estadístico chi-cuadrado
chi_cuadrado_u = np.sum((observados- esperados) ** 2 / esperados)

# Calcular los grados de libertad
df_u = len(terremotos_b100k_m2000t['mes_acumulado']) - 1 - 2  # Ajustar según los parámetros estimados

# Calcular el valor crítico
alpha = 0.05  # Nivel de significancia
critical_value = chi2.ppf(1 - alpha, df_u)

# Comparar estadístico con el valor crítico
if chi_cuadrado_u > critical_value:
    print("Rechazamos la hipótesis nula")
else:
    print("No podemos rechazar la hipótesis nula")
```

```python


# Valores observados
valores_observados_t = chile['depth']

# Calcular valores esperados usando la pdf de la distribución t de Student
valores_esperados_u = t.pdf(valores_observados_t, 3, 30, 16)

# Convertir a array para el cálculo del estadístico chi-cuadrado
observados2 = np.array(valores_observados_t)
esperados2 = np.array(valores_esperados_u)

# Calcular el estadístico chi-cuadrado
chi_cuadrado_u = np.sum((observados2 - esperados2) ** 2 /esperados2)

# Calcular los grados de libertad
df_u = len(valores_observados_t) - 1 - 3  # Ajustar según los parámetros estimados (3 parámetros en la t de Student)

# Calcular el valor crítico
alpha = 0.05  # Nivel de significancia
critical_value = chi2.ppf(1 - alpha, df_u)

# Comparar estadístico con el valor crítico
if chi_cuadrado_u > critical_value:
    print("Rechazamos la hipótesis nula")
else:
    print("No podemos rechazar la hipótesis nula")
```
